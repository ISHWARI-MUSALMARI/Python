```python
#print even number from 1 to 15 using while loop
i=0
while i<15:
    print(i)
    i+=2
```

```python
#print odd number from 1 to 15 using while loop
i=1
while i<15:
    print(i)
    i+=2
```

```python
#while loop has 2 more loops
#continue 
#skip the specific loop and then continue
i=0
while i<5:
    i=i+1
    if i==3:
        continue
    print(i)
```

```python
#break
#exit the loop
i=0
while i<5:
    i=i+1
    if i==3:
        break
    print(i)
```

```python
#code to extract all even number from the list (numfor )
num =[1,2,3,4,5,6,7,8,23,46,67,89]
for i in num:
    if i%2==0:
        print(i)
    
```

```python
#code to extract all odd number
num =[1,2,3,4,5,6,7,8,23,46,67,89]
for i in num:
    if i%2==1:
        print(i)
```

```python
#both even and odd num
num =[1,2,3,4,5,6,7,8,23,46,67,89]
for i in num:
    if i%2==0:
         print(i,"even num")
    else:
        print(i,"odd num")
    
        
```
