```python
#multiple nestedif condition
nationality =input("Enter your nationality")
age=int(input("Enter your age"))
if nationality.lower()=="indian"  or nationality.upper()=="INDIAN":
        print("citizen")
        if age>=18:
            print("eligible for vote")
        else:
            print("underage")
else:
        print("not a citizen")
```

```python
#bank loan approval
country = input("Enter your country:")
employment = input("Enter your employment status:")
salary = int(input("Enter your salary: "))
credit_score = int(input("Enter your credit score:"))
if country.lower() == "india" or country.upper()=="India":
    if employment.lower() == "employed" or employment.upper()=="Employed":
        if salary >= 50000:
            if credit_score >= 900:
                print("Loan Approved")
            else:
                print("Loan Rejected")
        else:
            print("Employed")
    else:
        print("Not Employed")
else:
    print("Loan Rejected")
```

```python
#college admisiion
state = input("Enter your state")
marks=int(input("Enter your marks"))
age=int(input("Enter your age"))
qualification =input("Enter your qualification")
if state.lower()=="karnataka" or state.upper()=="Karnataka":
    if age>=18:
        if marks>=50:
            print("approved")
        else:
            print("not approved")
    else:
        print("underage")
else:
    print("not qualified")
```

```python
#online shopping
login = input("Are you logged in?")
product = input("Is product available?")
payment = input("Payment successful?")
if login.lower() =="yes"  or login.upper()=="no":
    if product.lower()=="yes" or product.upper()=="no":
        if payment.lower() == "yes" or payment.upper()=="no":
            print("Order confirmed")
        else:
            print("Payment failed")
    else:
        print("Product not available")
else:
    print("Please login first")
```

```python
#ATM withdrawal
card = input("Is your Card valid:")
pin = int(input("Enter your PIN:"))
balance = int(input("Enter your Balance amount:"))
amount = int(input("Amount to withdraw:"))
if card.lower() == "yes":
    if pin == 1234:
        if balance >= amount:
            print("Withdraw successful")
        else:
            print("Insufficient balance")
    else:
        print("Wrong PIN")
else:
    print("Invalid card")
    
```

```python
#insurance eligiblity
country = input("Enter your Country:")
age = int(input("Enter your Age:"))
income = int(input("Enter your montly Income:"))
medical = input("Medical criteria:")

if country.lower()=="india" or country.upper()=="India":
    if age >= 18:
        if income>=30000:
            if medical.lower()=="yes" or medical.upper()=="no":
                print("Insurance Approved")
            else:
                print("Medical criteria not met")
        else:
            print("Low income")
    else:
        print("Underage")
else:
    print("Country not eligible")
```

```python
#employee promotion
department = input("Enter your Department name:")
experience = int(input("Experience time period:"))
performance = int(input("Performance in the company:"))
salary = int(input("annual Salary: "))
if department.lower() == "sales" or department.upper()=="Sales":
    if experience>=3:
        if performance>=80:
            if salary>=30000:
                print("Promotion Approved")
            else:
                print("Salary is very low")
        else:
            print("Low performance in the company")
    else:
        print("Less experience")
else:
    print("Department not eligible")
```
